## **Busca Binária (Binary Search)**

O algoritmo de busca binária é utilizado para encontrar um elemento específico em uma lista de elementos ordenada.

A lista é dividida ao meio a cada iteração; com a estratégia de divisão e conquista, a complexidade do pior caso é:
$$O(log(n))$$

```python
def binary_search(arr: list, start: int, end: int, key):

    if start>end:
        return None

    mid = (start + end) // 2
    if arr[mid] == key:
        return mid
    
    if key < arr[mid]:
        return binary_search(arr, start, mid-1, key)

    return binary_search(arr, mid + 1, end, key)
```

```python
arr = [i for i in range(20)]
print(arr)
```

```python
x1 = 13
print(binary_search(arr, 0, len(arr) - 1, x1))
x2 = 99
print(binary_search(arr, 0, len(arr) - 1, x2))
x3 = 1
print(binary_search(arr, 0, len(arr) - 1, x3))
```

## **Busca Linear**

É a forma mais simples de procurar um elemento em uma lista. A lista é percorrida item por item, do início ao fim, comparando cada elemento com o valor procurado.

Esse algoritmo não exige uma lista ordenada e sua complexidade, no pior caso, é:

$$O(n)$$

```python
import random as rd
```

```python
def linear_search(arr, key):

    for i in range(len(arr)):
        if arr[i] == key:
            return i
    return None
```

```python
arr = [rd.randint(1, 20) for i in range(20)]
print(arr)
```

```python
x1 = 13
print(linear_search(arr, x1))
x2 = 10
print(linear_search(arr, x2))
x3 = 1
print(linear_search(arr, x3))
```
