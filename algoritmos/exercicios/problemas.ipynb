### **Problema 1**

```python
def func(n):
    i = s = 1
    while s <= n:
        print ( i )
        print ( s )
        i = i + 1
        s = s + i

func(40)
```

### **Solução:**

$$T(n) = O(\sqrt n)

### **Problema 2**

```python
def func (n):

    i = 1

    count = 0

    while i*i <= n:

        print(i)

        print(count)

        i+=1

        count+=1
        
func(40)
```

### **Solução:**

$$T(n) = O(\sqrt n)

### **Problema 3**

```python
def func(n):

    count = 0

    for i in range(n // 2, n): # n/2

        j = 1

        while j + (n//2) <= n: # n/2

            k = 1

            while k <= n: # log(n)

                count += 1
                k *= 2

            j += 1

    print(count)

func(40)
```

### **Solução**:

$$T(n) = O(n²log(n))$$

### **Problema 4**

```python
def func(n):

    count = 0

    for i in range(n//2, n):

        j = 1

        while j + (n//2) <= n:

            break

        j *= 2

    print(count)

func(40)
```

### **Solução**:

$$ T(n) = O(n) $$
