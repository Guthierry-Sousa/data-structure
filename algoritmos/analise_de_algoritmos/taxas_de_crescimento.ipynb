## **Taxas de Crescimento**

```python
import numpy as np
import matplotlib.pyplot as plt
import random as rd
```

### **Complexidade Constante (1)**

```python
def imprimir_primeiro_elemento(data):

    print(data[0]) # Complexidade constante (1)

string = "Guthierry"

lista = [19, 10, 7, 8]

imprimir_primeiro_elemento(string)
imprimir_primeiro_elemento(lista)
```

### **Complexidade Linear (n)**

```python
def busca_linear(data, item):

    for i in range(len(data)): # Complexidade Linear (n)

        if data[i] == item:

            return i

    return None
```

### **Complexidade Logarítmica (log n)**

```python
def busca_binaria(data, item): # Complexidade logarítmica (log n)

    start = 0
    end = len(data) - 1

    while start <= end:

        middle = (start + end) // 2

        if data[middle] == item:

            return middle

        elif item > data[middle]:

            start = middle + 1

        else:

            end = middle - 1

    return None
        
lista = [1, 2, 3, 4 ,5]

print(busca_binaria(lista, 5))

print(busca_binaria(lista, 19))

print(busca_binaria(lista, 1))
```

### **Complexidade Quadrática (n²)**

```python
def selection_sort(data): # Complexidade quadrática (n²)

    n = len(data)

    for i in range(n-1):

        min = i

        for j in range(i+1, n):

            if data[j] < data[min]:

                min = j

        if min != i:

            data[i], data[min] = data[min], data[i]

    return data

lista = [rd.randint(1, 100) for _ in range(20)]

print(selection_sort(lista))
```

### **Quicksort (n log n)**

```python
def quick_sort(data): # Complexidade (n log n) -> caso médio

    if len(data) < 2:

        return data

    pivot = data[0]

    lower = [i for i in data[1:] if i <= pivot]
    upper = [i for i in data[1:] if i > pivot]

    return quick_sort(lower) + [pivot] + quick_sort(upper)

lista = [rd.randint(1, 100) for _ in range(20)]

print(quick_sort(lista))
```

### **Gráfico das taxas de crescimento**

```python
n = np.arange(1, 7)
```

```python
def constante(n):

    return np.ones(n.shape)

def logaritmica(n):

    return np.log2(n)

def linear(n):

    return n

def linear_logaritmica(n):

    return n * np.log2(n)

def quadratica(n):

    return n ** 2

def exponencial(n):

    return 2 ** n
```

```python
plt.figure(figsize=(10,6))
plt.title("Taxas de Crescimento (Complexidade)")

plt.plot(n, constante(n), color = 'red', label = "1")
plt.plot(n, logaritmica(n), color = 'green', label = "log n")
plt.plot(n, linear(n), color = 'blue', label = "n")
plt.plot(n, linear_logaritmica(n), color = 'orange', label = "n log n")
plt.plot(n, quadratica(n), color = 'purple', label = "n²")
plt.plot(n, exponencial(n), color = 'yellow', label = "2^n")

plt.ylabel("Tempo de Execução")

plt.grid(alpha = 0.3)

plt.legend()

plt.show()
```
