## **Comparação de Algoritmos**

Medir diretamente o tempo de execução em um computador, não é ideal, pois depende do **hardware**, do **compilador** e da **linguagem de programação**, dificultando comparações justas.

A abordagem mais confiável é descrever o tempo de execução em função do tamanho da entrada: $f(n)$.

Para valores muito grandes de n, **termos de menor ordem e constantes tornam-se irrelevantes**.

### **Exemplo**

$$T1(n) = n⁴ + 2n² + 10n + 650$$ e $$T2(n) = n⁴$$

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def t1(n):

    return n**4 + 2*n**2 + 10*n + 650

def t2(n):

    return n**4 

n = np.arange(1, 1001)

y1 = t1(n)

y2 = t2(n)
```

```python
print(f"T1(n) para n = 1000: {y1[-1]}")
print(f"T2(n) para n = 1000: {y2[-1]}")
```

```python
fig, axs = plt.subplots(1, 3, figsize = (15, 5))

axs[0].plot(n, y1, color = "red", alpha = 0.5)
axs[0].set_title("Função Completa")
axs[1].plot(n, y2, color = "blue", alpha = 0.5)
axs[1].set_title("Apenas o termo dominante")
axs[2].plot(n, y1, color = "red", alpha = 0.5, linestyle = "--", label = f"$T(n) = n⁴ + 2n² + 10n + 650$")
axs[2].plot(n, y2, color = "blue", alpha = 0.5, label = f"$T(n) = n²$")
axs[2].set_title("Comparação")
axs[2].legend()

plt.show()
```

É possível verificar que, para valores altos de $n$, é praticamente impossível notar a diferença entre as duas funções.

### **Crescimento Quadrático**

$$f(n) = n²$$ e $$g(n) = n² - 3n + 2$$

```python
def f(n):

    return n*n

def g(n):

    return n*n - 3*n + 2

print(f"f(n) para n = 1000: {f(1000)}")
print(f"g(n) para n = 1000: {g(1000)}")
print(f"Diferença: {(f(1000) - g(1000))/f(1000)}")
```

A diferença relativa entre elas é inferior a 3%. Assim, para valores grandes de n, ambas as funções apresentam comportamento semelhante. As constantes e os termos de menor grau não fazem muita diferença.

```python
fig, axs = plt.subplots(1, 3, figsize = (15, 5))

axs[0].plot(n[:10], f(n[:10]), color = "red", alpha = 0.5)
axs[0].plot(n[:10], g(n[:10]), color = "blue", alpha = 0.5)
axs[0].set_title("n próximo de 0")

axs[1].plot(n[:100], f(n[:100]), color = "red", alpha = 0.5)
axs[1].plot(n[:100], g(n[:100]), color = "blue", alpha = 0.5)
axs[1].set_title("n próximo de 100")

axs[2].plot(n, f(n), color = "red", alpha = 0.5, label = f"$f(n) = n²$")
axs[2].plot(n, g(n), color = "blue", alpha = 0.5, label = f"$f(n) = n² - 3n + 2$")
axs[2].set_title("Comparação Final")
axs[2].legend()

plt.show()
```

### **Conclusão**

Esses exemplos mostram que funções diferentes podem ter comportamento semelhante quando $n$ é muito grande. Por isso, a função de custo do algoritmo é associada a uma **taxa de crescimento dominante**, desconsiderando constantes e termos de menor ordem.
