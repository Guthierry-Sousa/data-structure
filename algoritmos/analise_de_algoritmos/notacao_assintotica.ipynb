## **Notação Assintótica**

A notação assintótica permite analisar o comportamento do algoritmo quando o tamanho da entrada tende a valores muito grandes. Ela serve para determinar limites superiores e inferiores de tempo de execução.

### **Notação Big O**

A notação big O determina o **limite superior** do tempo de execução, ou seja, normalmente esta associada ao pior caso do algoritmo. Essa é a notação mais usada na prática, pois é útil garantir o pior caso de um algoritmo.

$$f(n)=O(g(n)) \iff \exists\,c>0,\;n_0>0:\;0\leq f(n)\leq c\,g(n),\;\forall n\geq n_0$$

A partir de algum tamanho $n_0$, $f(n)$ nunca cresce mais rápido que uma constante vezes $g(n)$.

**Exemplo 1**: 

$$f(n) = n⁴ + 2n² + 10n + 650$$

$$f(n) = O(n⁴)$$

Pois, para $n\geq 1$ : $n⁴ + 2n² + 10n + 650 \leq 663n⁴$

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(1, 1001)
```

```python
def f(n):

    return n**4 + 2*n**2 + 10*n + 650

def g(n):

    return n**4

y1 = f(n)
y2 = g(n)

plt.figure(figsize=(8,5))

plt.plot(n, y1, color = 'blue', label = "f(n)")
plt.plot(n, y2, color = 'red', linestyle = '--', label = 'g(n) = n⁴')
plt.ylabel("Taxa de Crescimento")
plt.xlabel("n")

plt.grid(alpha = 0.4)
plt.legend()
plt.show()
```

### **Notação Ômega**

A notação $\Omega$ determina o **limite inferior** do tempo de execução. Essa notação é usada tipicamente para expessar o melhor caso de um algoritmo, ou para definir limites inferiores de complexidade.

$$f(n)=\Omega(g(n)) \iff \exists\,c>0,\;n_0>0:\;0\leq c\,g(n)\leq f(n),\;\forall n\geq n_0$$

A partir de algum ponto $n_0$, $f(n)$ é sempre maior ou igual a uma constante vezes $g(n)$.

**Exemplo 2**:

$$f(n) = 100n³ + 100n² + 50$$

$$f(n) = \Omega(n³)$$

Pois, para $n\geq 0$ : $100n³ + 100n² + 50 \geq 100n³$

```python
def f(n):

    return 100*n**3 + 100*n**2 + 50

def g(n):

    return n**3

y1 = f(n)
y2 = g(n)

plt.figure(figsize=(8,5))

plt.plot(n, y1, color = 'blue', label = "f(n)")
plt.plot(n, y2, color = 'red', linestyle = '--', label = 'g(n) = n³')
plt.ylabel("Taxa de Crescimento")
plt.xlabel("n")

plt.grid(alpha = 0.4)
plt.legend()
plt.show()
```

### **Notação Theta**

A notação $\Theta$ é usada quando o limite inferior e superior de um algoritmo são os mesmos.

$$f(n)=\Theta(g(n)) \iff \exists\,c_1,c_2>0,\;n_0>0:\;0\leq c_1g(n)\leq f(n)\leq c_2g(n),\;\forall n\geq n_0$$

A partir de algum ponto $n_0$, $f(n)$ fica presa entre duas constantes vezes $g(n)$.

**Exemplo 3**:

$$f(n) = 30n² + 100n + 10$$

$$f(n) = O(n²)$$
$$f(n) = \Omega(n²)$$

Logo, $$f(n) = \Theta(n²)$$

```python
def f(n):

    return 30*n**2 + 100*n + 10

def g(n, c = 1):

    return c*n**2

y1 = f(n)
y2 = g(n, 140)
y3 = g(n)

plt.figure(figsize=(8,5))

plt.plot(n, y1, color = 'blue', label = "f(n)")
plt.plot(n, y2, color = 'red', linestyle = '--', label = 'c1 * g(n) = 140n²')
plt.plot(n, y3, color = 'green', linestyle = '--', label = 'c2 * g(n) = 1n²')
plt.ylabel("Taxa de Crescimento")
plt.xlabel("n")

plt.grid(alpha = 0.4)
plt.legend()
plt.show()
```

### **Resumo**



A análise assintótica estuda o comportamento dos algoritmos à medida que o tamanho da entrada n cresce, permitindo comparações independentes do hardware e do ambiente de execução.
