## **POO em Python - Classes e Objetos**

Em Python, tudo é orientado a objetos, e as classes são a base para a criação de objetos. Classes permitem agrupar dados (atributos) e comportamentos (métodos) um uma única estrutura.

### **Classes e Objetos**

Uma **classe** é uma estrutura que define atributos (características) e métodos (comportamentos) que os objetos criados a partir dela vão ter.

Um **objeto** é uma instância concreta de uma classe.

```python
class Cachorro:

    def __init__(self, nome, raca):
        self.nome = nome
        self.raca = raca

c = Cachorro("Rex", "PitBul")

print(c.nome)
print(c.raca)
```

O método **__init__** é um construtor. Ele é utilizado para inicializar/definir o estado interno do objeto no momento da sua criação. Este método é chamado automaticamente quando uma classe é instânciada.


O **self** representa a própia instância que esta sendo criada ou manipulada.

```python
class Cachorro:

    def __init__(self, nome, raca):
        self.nome = nome # Atributo
        self.raca = raca # Atributo

    def latir(self): # Método
        print(f"{self.nome} está latindo: AuAuAu!")


c = Cachorro("Rex", "PitBul")

c.latir()
```

### **Atributos de Classe x Atributos de Instância**


**Atributos de Classe**: Existem em um único local na memória e são compartilhados por todas as instâncias (objetos)

**Atributos de Instância**: São únicos para cada objeto criado.

```python
class Pessoa:

    especie = 'Homo Sapiens' # Atributo de Classe

    def __init__(self, nome, idade):
        self.nome = nome # Atributo de Instância
        self.idade = idade # Atributo de Instância

p1 = Pessoa('Guthy' , 19)
p2 = Pessoa('Elis', 19)

print(p1.especie)
print(p2.especie)
print(p1.nome)
print(p2.nome)
```

### **Herança**

Permite que uma classe reutilize e estenda o comportamento de outra classe.

```python
class Animal:

    def __init__(self, nome):
        self.nome = nome

    def emitir_som(self):
        print(f"{self.nome} faz som!")


class Gato(Animal):

    def __init__(self, nome, cor_pelo):
        super().__init__(nome)
        self.cor_pelo = cor_pelo

    def emitir_som(self): # sobrescreve o método da classe pai
        print(f"{self.nome} faz som: Miauuuuu")

class Cachorro(Animal):

    def __init__(self, nome, raca):
        super().__init__(nome)
        self.raca = raca

    def emitir_som(self): # sobrescreve o método da classe pai
        print(f"{self.nome} faz som: AuaAuAu")


animais = [Gato("Misha", "Branco"), Cachorro("Totti", "PitBull")]

for animal in animais:
    animal.emitir_som() # Polimorfismo: Objetos diferentes podem responder ao mesmo método de formas diferentes
```
