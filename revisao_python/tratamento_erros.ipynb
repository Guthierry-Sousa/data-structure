## **Tratamento de Erros**

### **Exceção**

Uma exceção é um evento que interrompe o fluxo normal de execução de um programa quando ocorre um erro.

```python
num = int('a') # Gera uma exception e exibe o traceback
```

### **Estrutura: Try / Except**

```python
try:

    a = int(input("Informe um número: "))

    b = int(input("Informe outro número: "))

    d = a / b

except ValueError as e:

    print("Valor Inválido")
    print(f"Erro: {e}")

except ZeroDivisionError as e:

    print(f"Erro: {e}")
```

### **Else e Finally**

**Else**: Executa somente se nenhuma exceção ocorreu no bloco try.

**Finally**: Executa sempre, independente de ter ocorrido erro ou não.

```python
try:

    a = int(input("Informe um número: "))

    b = int(input("Informe outro número: "))

    d = a / b

except ValueError as e:

    print("Valor Inválido")
    print(f"Erro: {e}")

except ZeroDivisionError as e:

    print(f"Erro: {e}")

else:

    print(f"{a}/{b} = {d}")

finally: 

    print("Execução Finalizada!")
```

### **Gerando exceções manualmente (raise)**

```python
def imprimir_idade(idade: int):

    if idade < 0:

        raise ValueError("Idade não pode ser negativa!")

    print(f"Idade = {idade}")

imprimir_idade(-10)
```

Em Python todas as exceções derivam de uma classe base chamada **BaseException**.

Mas, a maioria dos erros mais comuns herdam de **Exception**.



```python
class NotaForaDoIntervalo(Exception):
    pass

def cadastrar_nota():

    while True:

        try:

            nota = float(input("Informe a nota: "))

            if nota < 0 or nota > 10:

                raise NotaForaDoIntervalo("Erro: Nota fora do intervalo permitido (0, 10).")

        except ValueError:

            print("Erro: Número inválido")

        except NotaForaDoIntervalo as e:

            print(e)

        else:

            print(f"Nota Cadastrada: {nota}")
            break

        finally:

            print("Tentativa de Cadastro Finalizada!")

cadastrar_nota()
              
```
