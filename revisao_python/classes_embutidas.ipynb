## **Classes Embutidas (Built-in) - Python**

### **Sequências**

As sequências são coleções ordenadas de itens. Cada elemento pode ser acessado por meio de um índice numérico, começando em zero.

#### **Listas (list)**

Uma lista é uma sequência ordenada e mutável de elementos. É possível adicionar, remover ou alterar elementos a qualquer momento.

 - Em Python as listas são implementadas como arrays dinâmicos.

- Diferente de arrays em C, uma lista Python pode misturar tipos, porque na verdade ela armazena referências para os objetos, não os valores diretamente.


```python
numeros = [x for x in range(10)]

numeros[0] = 'Maça'

numeros[-1] = [100, 101]

numeros.append(17)

numeros.pop()

numeros
```

```python
lista = [15, 8, 23, 4, 42, 16, 9]
lista.append(100)
lista.remove(8)
lista[1] = 50

lista_ordenada = sorted(lista)
print(lista_ordenada)

max_value = max(lista)
min_value = min(lista)
sum_elements = sum(lista)

print(f"Maior valor: {max_value}")
print(f"Menor valor: {min_value}")
print(f"Soma dos valores: {sum_elements}")
```

#### **Tuplas (tuple)**

Uma tupla é uma sequência ordenada e imutável de elementos.

- Não é possível alterar, adicionar e nem remover elementos.

- Tuplas são mais eficientes que listas, porque a estrutura interna não precisa reservar espaço extra para crescimento futuro.

```python
coordenadas = (10, 20)
pessoa = ("Guthy", 19, "Cientista da Computação")

print(coordenadas)
print(pessoa)
```

```python
print(pessoa[0])
```

```python
try:

    pessoa[2] = "Engenheiro"

except Exception as e:

    print(e)
```

```python
def dividir(a, b):

    return a//b, a%b

q, r = dividir(10, 4)

print(q)
print(r)
```

```python
tupla = ("Guthy", 7.5, 8.0, 6.5)

nome, nota1, nota2, nota3 = tupla

media = (nota1 + nota2 + nota3)/3

if media > 7.0:

    print(f"{nome} foi aprovado com média {media:.2f}")

else:

    print("REPROVADO!")
```

#### **Strings (str)**

Uma String é uma sequência ordenada e imutável de caracteres.

- Assim como as tuplas, strings não podem ser alteradas depois de criadas. Qualquer 'modificação' cria uma nova string na memória.

```python
string = 'Hello World'

string = string.upper() # Uma nova string é criada

print(string)
```

```python
frase = input("Informe uma frase: ")

print(f"Frase com letras maiúsculas: {frase.upper()}")
print(f"Frase com letras minúsculas: {frase.lower()}")
print(f"Número de caracteres: {len(frase)}")
print(f"Frase invertida: {frase[::-1]}")

if 'python' in frase.lower():

    print("A frase digitada possui a palavra 'Python'")

else:

    print("Palavra 'Python' não encontrada na frase")
```

### **Conjuntos (set)**

Um conjunto é uma coleção não ordenada de elementos únicos.

- Um set em Python é implementado como uma tabela hash (hash table).

- Ele possui todas operações características da teoria dos conjuntos

```python
conjunto = {1,2,3,3,3,3,4,4,1,1,5,5,6,7}
print(conjunto)
```

```python
try:

    print(conjunto[0])

except Exception as e:

    print(e)
```

```python
conjunto.add(100)
print(conjunto)
```

```python
fs = frozenset([1,2,3]) # Tipo de conjunto imutável

try:

    fs.add(100) 

except Exception as e:

    print(e)
```

```python
alunos_matematica = {'Ana', 'Bruno', 'Carla', 'Diego', 'Elena'}

alunos_portugues = {'Carla', 'Diego', 'Fernanda', 'Gustavo'}

alunos_matematica_e_portugues = alunos_matematica & alunos_portugues

alunos_matematica_ou_portugues = alunos_matematica | alunos_portugues

alunos_somente_matematica = alunos_matematica - alunos_portugues

print(f"Alunos matriculados em ambas as diciplinas: {alunos_matematica_e_portugues}")
print(f"Alunos matriculados em pelo menos uma das duas disciplinas: {alunos_matematica_ou_portugues}")
print(f"Alunos matriculados somente em Matemática: {alunos_somente_matematica}")
```

```python
def qtd_elementos_unicos(lista: list):

    return len(list(set(lista)))


lista = ["Ana", "Ana", "Bruno", "Carla", "Carla", "Carla"]

print(f'Existem {qtd_elementos_unicos(lista)} elementos únicos na lista.')
```

### **Dicionários (dict)**

Um dicionário associa chaves a valores. É a implementação de uma hashmap em Python.

- As chaves de um dicionário precisam ser únicas e imutáveis.

- O acesso é feito por chave, não por índice.

```python
aluno = {'idade': 19, 'nome': 'Guthierry', 'curso': 'Computação'}

print(len(aluno))
print(aluno['idade'])
print(aluno.keys())
print(aluno.values())
print(aluno.items())
```

```python
estoque = {'caneta': 50, 'lapis': 20, 'borracha': 100, 'lapiseira': 5, 'caderno': 1}

estoque['apontador'] = 17

estoque['caneta'] -= 5

del estoque['lapis']

estoque.get('mochila', 'produto não encontrado')

for p, qtd in estoque.items():

    print(f"Produto: {p} | Quantidade em estoque: {qtd}")
```

```python
sum_qtd_total = sum(k for k in estoque.values())
print(f'Quantidade total de produtos: {sum_qtd_total}')
```

#### **Dict comprehension**

```python
quadrados = {x: x*x for x in range(1,11)}

for k, v in quadrados.items():

    print(f"{k} ao quadrado = {v}")
```
