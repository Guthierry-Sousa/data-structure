## **Filas**

### **Implementação com array circular**

```python
from queue_array import QueueArray

q_array = QueueArray(3)
```

```python
print(q_array.is_empty())
```

```python
q_array.enqueue(1)
q_array.enqueue(2)
q_array.enqueue(3)
```

```python
print(q_array.is_full())
```

```python
try:

    q_array.enqueue(4)

except Exception as e:

    print(e)
```

```python
print(q_array.dequeue())
```

### **Implementação com lista encadeada**
