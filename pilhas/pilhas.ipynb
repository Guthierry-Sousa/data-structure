## **Pilhas (Stacks)**

Uma pilha é uma estrutura de dados que armazena os dados em uma ordem específica como ocorre nos arrays e nas listas encadeadas, com algumas restrições:

- Elementos de dados só podem ser inseridos no fim/topo da pilha (operação **push**)

- Elementos de dados só podem ser removidos no fim/topo da pilha (operação **pop**)

- Só o último elemento de dados pode ser consultado/lido na pilha (operação **peek**)

Uma pilha permite armazenar e ler dados em uma extremidade, e o elemento que é adicionado por último é o primeiro a ser selecionado. 

Logo, uma pilha é uma estrutura **LIFO (last in, first out**; último a entrar é o primeiro a sair).

### **Implementação com array estático**

A implementação com arrays utiliza um vetor de tamanho fixo para armazenar os elementos. 
O controle da posição do topo é feito por meio de um índice.

```python
from stack_arrays import StackArray
```

```python
pilha1 = StackArray(5)
```

```python
pilha1.push(1)
pilha1.push(2)
pilha1.push(3)

print(pilha1.is_empty())
```

```python
print(pilha1.peek())
```

```python
print(pilha1.pop())
```

```python
print(pilha1.peek())
```

```python
print(pilha1.pop())
print(pilha1.pop())

print(pilha1.is_empty())

try:

    pilha1.pop()

except IndexError as e:

    print(e)
```

```python
pilha1.push(1)
pilha1.push(2)
pilha1.push(3)
pilha1.push(4)
pilha1.push(5)

print(pilha1.is_full())

try:

    pilha1.push(6)

except IndexError as e:

    print(e)
```

### **Implementação com Array Dinâmico**

A implementação com arrays dinâmicos permite redimensionar o array quando a capacidade máxima é atingida. Isso torna a pilha mais flexível em relação ao crescimento e
redução de elementos.

```python
from stack_arrays import StackArrayDynamic
```

```python
pilha2 = StackArrayDynamic(5)
```

```python
for i in range(10):

    pilha2.push(i)

    print(pilha2.peek())

print(pilha2.pop())
```

```python
while True:

    try:

        print(pilha2.pop())

    except IndexError as e:

        print(e)
        break
```
