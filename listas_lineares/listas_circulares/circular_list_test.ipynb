## **Listas Circulares**

```python
from circular_list import CircularList
```

```python
def iter(lista: CircularList):

    if lista.head is None:
        return 

    current = lista.head

    while True:

        print(current.data)

        current = current.next

        if current == lista.head:
            break
```

```python
cir_list = CircularList()
```

### **Inserções**

```python
cir_list.insertion_at_beginning('Elis')
cir_list.insertion_at_beginning('Elen')
cir_list.insertion_at_beginning('Guthy')
```
