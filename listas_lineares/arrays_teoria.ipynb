## **Arrays - Listas Lineares com alocação sequencial**

Array é uma estrutura de dados em que seus elementos são **armazenados de maneira contígua/sequencial na memória**, isto é, um ao lado do outro.

Essa característica facilita o cálculo da posição de qualquer elemento do array com o uso de *endereços base* e de *offset*. O endereço base se refere ao endereço do local em que o primeiro elemento do array está armazenado, e o offset é um inteiro que indica o deslocamento do endereço base até determinado elemento.

A alocação de um array na memória pode ser:

- **Estática**: quando o tamanho é definido em tempo de compilação e é fixo.

- **Dinâmica**: quando o tamanho é definido em tempo de execução e é variável.

**Características**:

- Acesso Aleatório (rápido para busca);

- Não ideal para armazenamento de um volume muito grande de dados, pois será difícil encontrar blocos contíguos na memória;

- Difícil inserir e excluir elementos (precisa deslocar);

- Ideal quando se deseja fazer várias buscas e poucas inserções e exclusões.

```python
arr = [1, 2, 3] # -> Arrays em Python são dinâmicos

arr.pop(0)

arr.append(10)

print(arr)
```
