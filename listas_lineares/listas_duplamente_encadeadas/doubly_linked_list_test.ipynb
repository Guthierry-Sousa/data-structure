## **Listas Duplamente Encadeadas**

```python
from doubly_linked_list import DoublyLinkedList
```

```python
def iter(lista: DoublyLinkedList):

    current = lista.head

    while current:
        print(current.data)
        current = current.next
```

```python
words = DoublyLinkedList()
```

### **Inserção no início**

```python
words.insertion_at_beginning('Guthy')
words.insertion_at_beginning('Elis')
words.insertion_at_beginning('Elen')
```

```python
iter(words)
```

```python
print(words.head.data)
print(words.tail.data)
```

### **Inserção no final**

```python
words.insertion_at_end('Gleyci')
words.insertion_at_end('JP')
```

```python
iter(words)
```

### **Inserção em qualquer posição**

```python
words.insert_at_given_position('Dayane', 1)
words.insert_at_given_position('Edson', 3)
```

```python
iter(words)
```

### **Consulta**

```python
words.search('Elis')
```

```python
words.search('Luis')
```

```python
iter(words)
```
