## **Listas Encadeadas**

As listas encadeadas são estruturas de dados dinâmicas e lineares.

Os elementos de dados são **armazenados na memória em locais não contíguos** e são conectados por ponteiros (variável que armazena um endereço de memória). Cada elemento referencia (ou aponta) para o próximo elemento de dados até que seja alcançado o último elemento, que aponta para *None*.

Cada item de dados em uma lista encadeada se chama **Nó**. Um nó armazena os dados reais e um ponteiro para o próximo nó.

O armazenamento não contíguo facilita as operações de inserção e remoção, pois não é necessário deslocar os elementos.

**Características**:

- Boa para inserção e remoção, comparado aos arrays;

- O tamanho pode aumentar ou diminuir durante a execução do programa;

- Ideal para tarefas, no qual utilize poucas operações de busca e mais operações de inserção e remoção.

### **Implementação da classe Node (nó)**

```python
class Node:

    def __init__(self, data):
        self.__data = data # armazena o dado
        self.__next = None # armazena a referência para o próximo nó

    def get_data(self):
        return self.__data

    def set_data(self, value):
        self.__data = value

    def get_next(self):
        return self.__next

    def set_next(self, value):
        self.__next = value
```

### **Implementação Simples**

```python
n1 = Node('guthy')
n2 = Node('elis')
n3 = Node('elen')
n4 = Node('ramon')
```

```python
n1.set_next(n2)
n2.set_next(n3)
n3.set_next(n4)
```

```python
# Percorrendo uma lista encadeada

current = n1

while current != None:
    print(current.get_data())
    current = current.get_next()
```

```python
class SinglyLinkedList:

    def __init__(self):
        self.__head = None
        self.__size = 0

    def iter(self):

        if self.__size > 0:
            current = self.__head
            while current:
                val = current.get_data()
                current = current.get_next()
                yield val

    def insert_at_end(self, value): # inserir elementos no final -> O(n)

        node = Node(value)

        if self.__head is None:
            self.__head = node

        else:
            current = self.__head
            while current.get_next():
                current = current.get_next()
            current.set_next(node)

        self.__size += 1

    # def append(self, value): # inserir elementos no final -> O(1)

    #     node = Node(value)

    #     if self.__tail:
    #         self.__tail.set_next(node)
    #         self.__tail = node

    #     else:
    #         self.__head = node
    #         self.__tail = node

    #     self.__size += 1

    def insert_at_beginning(self, value): # inserir elementos no início
    
            node = Node(value)
    
            if self.__head is None:
                self.__head = node
    
            else:
                node.set_next(self.__head)
                self.__head = node

            self.__size += 1

    def insert_at_given_position(self, value, index): # Inserir em uma posição específica

        if (index < 0) or (index > self.__size):
            raise ValueError("Índice inválido!")

        if index == 0:
            self.insert_at_beginning(value)

        elif index == self.__size:
            self.insert_at_end(value)

        else:
            node = Node(value)
            current = self.__head
            count = 1
            while count < index:
                current = self.__head.get_next()
                count += 1

            node.set_next(current.get_next())
            current.set_next(node)
            self.__size += 1

    def del_from_beginning(self):

        if self.__head is not None:
            self.__head = self.__head.get_next()
            self.__size -= 1

    def del_from_end(self):

        if self.__head is None:
            return

        if self.__size == 1:
            self.del_from_beginning()

        else:

            current = self.__head
            previous = None

            while current.get_next():

                previous = current
                current = current.get_next()

            previous.set_next(None)
            self.__size -= 1

    def del_at_position(self, index):

        if (index < 0) or (index > self.__size):
            raise ValueError("Índice inválido!")

        if index == 0:
            self.del_from_beginning()

        elif index == self.__size:
            self.del_from_end()

        else:

            current = self.__head
            previous = None
            count = 1

            while count <= index:
                previous = current
                current = current.get_next()
                count += 1

            previous.set_next(current.get_next())

            self.__size -= 1
            

    def search(self, value): # Buscar elementos

        for node in self.iter():
            if node == value:
                return True

        return False

    def size(self): # Tamanho da lista
        return self.__size

    def clear(self):
        self.__head = None
        self.__size = 0
```

```python
names = SinglyLinkedList()
```
