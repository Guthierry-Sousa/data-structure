## **Árvores Binárias de Busca (Binary Search Tree)**

Uma **Árvore Binária de Busca** é uma árvore binária em que os valores de qualquer nó na árvore é maior que os valores de todos os nós de sua subárvore esquerda e menor aos valores de todos os nós da subárvore direita.

### **Operações em árvores binárias de busca**

```python
class Node:

    def __init__(self, key = None, data = None):
        self.data = data
        self.key = key
        self.left = None
        self.right = None
```

#### **Busca de Nós**

- se o valor for igual, a busca termina;

- se for menor, seguimos para a esquerda;

- se for maior, seguimos para a direita;



```python
def search(root: Node, key: int): 

    if root is None:
        return None

    if root.key == key:
        return root

    if key < root.key:

        if root.left:
            return search(root.left, key)

    if key > root.key:

        if root.right:
            return search(root.right, key)

    return root
```

```python
def search_01(root: Node, key):

    if root is None:
        return None

    if root.key == key:
        return root

    if key < root.key:
        return search_01(root.left, key)

    return search_01(root.right, key)
```

```python
def search_bool(root: Node, key: int) -> bool:

    if root is None:
        return False

    if root.key == key:
        return True

    if key < root.key:
        return search_bool(root.left, key)

    if key > root.key:
        return search_bool(root.right, key)
```

#### **Inserção**

```python
def insert_01(root: Node, node: Node):

    if root is None:
        root = node
        return root

    n = search(root, node.key)

    if n.key == node.key:
        n = node

    if node.key < n.key:
        n.left = node

    else:
        n.right = node
```

```python
def insert_02(root: Node, key):

    if root is None:
        return Node(key)

    if key < root.key:
        root.left = insert_02(root.left, key)

    if key > root.key:
        root.right = insert_02(root.right, key)

    return root
```

```python
def insert(root: Node, node: Node):

    if root is None or node.key == root.key:
        root = node
        return root

    current = root
    prev = None

    while current:

        if current.key == node.key:
            return root

        prev = current

        if node.key < current.key:
            current = current.left


        else:
            current = current.right

    if node.key < prev.key:
        prev.left = node

    else:
        prev.right = node

    return root
```
