## **Lista De Exercícios - Teoria**

### **1. O que é uma árvore binária própria?**

Um **árvore binária própria** é uma árvore binária em que todo nó tem exatamente 0 ou 2 filhos. Ou seja, nenhum nó possui apenas um filho.



### **2. O que é uma árvore binária cheia?**

Uma **árvore binária cheia** é uma árvore binária em que todos os nós internos têm exatamente dois filhos e todas as folhas estão no mesmo nível (mesma profundidade).



### **3. O que é uma árvore binária completa?**

Uma **árvore binária completa** é uma árvore binária em que todos os níveis estão totalmente preenchidos, exceto possívelmente o último nível. Caso o último nível não esteja totalmente preenchido, os nós que pertencem a esse nível ficam o mais à esquerda possível.




### **4. Qual a altura máxima de uma árvore binária com $n$ nós?**

A altura máxima de uma árvore binária com n nós é $n-1$ ou $n$, dependendo da definição utilizada. Isso ocorre em árvores degeneradas (cada nó possui exatamente um único filho).



### **5. Qual a altura mínima de uma árvore binária com $n$ nós?**

A altura mínima de uma árvore binária com n nós é $ h_{min} = \lfloor \log_2 n \rfloor + 1 $ ou $ h_{min} = \lfloor \log_2 n \rfloor $. Isso ocorre quando a árvore binária é perfeitamente balanceada, ou seja, para cada nó, o número de nós em suas subárvores esquerda e direita diferem no máximo em 1.



### **6. Qual é o número máximo e mínimo de nós internos e externos de uma árvore binária imprópria?**

Uma **árvore binária imprópria** é uma árvore binária em que, pelo menos 1 nó, possui exatamente 1 filho.



**Número Mínimo**

- Nós Externos: $1$ (Quando a árvore é degenerada)
- Nós Internos: $\lfloor n/2 \rfloor$.

**Número Máximo**

- Nós Externos: $\lfloor n/2 \rfloor$
- Nós Internos: $n-1$ ( Quando a árvore é degenerada)

### **7. Qual é o número mínimo de nós externos de uma árvore binária própria de altura h?**

$h+1$

### **8. Qual é o número máximo de nós externos de uma árvore binária própria de altura h?**

$2^h$. Ocorre quando a árvore binária é uma árvore do tipo cheia, no qual todos os níveis são preenchidos e todos os nós internos tem exatamente dois filhos.

### **9. O que é uma árvore binária balanceada?**

Uma árvore binária balanceada é aquela em que, para todo nó, a diferença entre as alturas das subárvore esquerda e direita é de no máximo 1.



### **10. O que é uma árvore binária perfeitamente balanceada?**

Uma árvore binária perfeitamente balanceada é aquela em que, para todo nó, o número de nós da subárvore esquerda e o da subárvore direita diferem em no máximo 1.
