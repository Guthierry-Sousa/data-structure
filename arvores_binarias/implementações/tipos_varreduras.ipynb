## **Varreduras em Árvores**

O método utilizado para a visita a todos os nós de uma árvore chama-se **varredura da árvore**.

### **Varredura em ordem (in-order tree travesal)**

Na varredura em ordem, visitamos os nós na ordem **subárvore esquerda**, **raiz** e **subárvore direita**.

```python
def in_order(root_node: Node):
    current = root_node

    if current is None:
        return

    in_order(current.left_child)
    print(current.data)
    in_order(current.right_child)

in_order(n0)
```

### **Varredura pré-ordem**

Na varredura pré-ordem, visitamos a árvore na ordem **nó raiz**, **subárvore esquerda** e **subárvore direita**.

```python
def pre_order(root_node: Node):

    current = root_node

    if current is None:
        return

    print(current.data)
    pre_order(current.left_child)
    pre_order(current.right_child)

pre_order(n0)
```

### **Varredura pós-ordem**

Na varredura pós-ordem, visitamos os nós na ordem **subárvore esquerda**, **subárvore direita** e **nó raiz**.

```python
def pos_order(root_node: Node):

    current = root_node

    if current is None:
        return

    pos_order(current.left_child)
    pos_order(current.right_child)
    print(current.data)

pos_order(n0)
```

### **Varredura em ordem de nível (level-order traversal)**

Nesse método de varredura, começamos visitando a raiz da árvore antes de visitar cada nó do próximo nível. Em seguida, passamos para o próximo nível da árvore e assim por diante.

obs: busca em largura (Breadth-First Search - BFS)

```python
from collections import deque

def level_order_traversal(root_node: Node):

    list_of_nodes = []

    traversal_queue = deque([root_node])

    while len(traversal_queue) > 0:

        node = traversal_queue.popleft()
        list_of_nodes.append(node.data)

        if node.left_child:
            traversal_queue.append(node.left_child)

            if node.right_child:
                traversal_queue.append(node.right_child)

    return list_of_nodes
```

```python
print(level_order_traversal(n0))
```
